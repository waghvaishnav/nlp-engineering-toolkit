# Feed Tokens to the Model :

explains how a pipeline works.

```python
from transformers import AutoTokenizer,AutoModelForSequenceClassification

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)
tokenizer = AutoTokenizer.from_pretrained(checkpoint)

text = [
    "work hard everyday !",
    "never give up until the fight ends !",
    "you are very shy !"
]
```

```python
# text converted into tokens .

tokens = tokenizer(text,return_tensors="pt",padding="max_length",max_length=10,truncation=True)

tokens
```

```python
# tokens provide or feed to the model.

output = model(**tokens)

output
```

```python
logits = output.logits

logits
```

post processing on logits :

```python
import torch
import torch.nn.functional as F

prob = F.softmax(logits,dim=1)

prob
```

```python
predicted_classes = torch.argmax(prob,dim=1).tolist()

predicted_classes
```

# Text >> Tokenizer >> Tokens(token_ids) >> Model >> Logits >> Post-Processing >> Output Text.
